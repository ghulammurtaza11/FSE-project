```python
import tkinter as tk
from tkinter import messagebox
import webbrowser
from IPython.display import display
from PIL import Image, ImageTk # murti pip intall pillow ker ok phir image load hogi.

# ================
# DASHBOARD PAGES
# ================
def show_student_dashboard(student_id, login_open_func):
    dash = tk.Toplevel(root)
    dash.title("Student Dashboard")
    dash.geometry("400x350")
    dash.configure(bg="#1c1c28")
    tk.Label(dash, text=f"Welcome, {student_id}!", bg="#1c1c28", fg="white", font=("Arial", 18)).pack(pady=20)

    def open_book_event():
        webbrowser.open("https://docs.google.com/forms/d/1Htgpiv5OCNNzripuSjGACE2BgICi7dZDL_UnkFj-XlE/edit")

    def open_our_work():
        our_work_window = tk.Toplevel(root)
        our_work_window.title("Our Work")
        our_work_window.geometry("600x600")
        our_work_window.configure(bg="#1c1c28")

        try:
            img = Image.open("uni.jpg")  #idher apni image ka path dale ga, same folder mein rakhna jisme ye code hai.
            img = img.resize((500, 500), Image.LANCZos)  #resize  image
            photo = ImageTk.PhotoImage(img)
            label = tk.Label(our_work_window, image=photo, bg="#1c1c28")
            label.image = photo  
            label.pack(pady=20)
            tk.Label(our_work_window, text="Description of the project or event.", bg="#1c1c28", fg="white").pack(pady=10)

        except Exception as e:
            messagebox.showerror("Error", f"Failed to load image: {e}")

    def open_socials():
        webbrowser.open("https://www.instagram.com/nu_vibin_?igsh=eWt4cDRueDUzd284")
    def logout():
        if messagebox.askyesno("Logout", "Are you sure you want to log out?"):
            dash.destroy()
            login_open_func()

    tk.Button(dash, text="BOOK EVENT", width=30, height=2, bg="#00bcd4", fg="white", command=open_book_event).pack(pady=10)
    tk.Button(dash, text="Our Work", width=30, height=2, bg="#00bcd4", fg="white", command=open_our_work).pack(pady=10)
    tk.Button(dash, text="Socials", width=30, height=2, bg="#00bcd4", fg="white", command=open_socials).pack(pady=10)
    tk.Button(dash, text="Logout", width=20, bg="#ff4d4d", fg="white", command=logout).pack(pady=20)


def show_admin_dashboard(login_open_func):
    dash = tk.Toplevel(root)
    dash.title("Admin Dashboard")
    dash.geometry("400x300")
    dash.configure(bg="#1c1c28")
    tk.Label(dash, text="Welcome, Admin!", bg="#1c1c28", fg="white", font=("Arial", 18)).pack(pady=20)

    def book():
        webbrowser.open("https://docs.google.com/spreadsheets/d/16mBMXr8xq8IK9KzwiHDWZ1JvWVx-xWmDuILtIpEf6T4/edit?resourcekey=&gid=1632270412#gid=1632270412")
    def view_requests():
        webbrowser.open("https://docs.google.com/spreadsheets/d/16mBMXr8xq8IK9KzwiHDWZ1JvWVx-xWmDuILtIpEf6T4/edit?resourcekey=&gid=1632270412#gid=1632270412")

    def logout():
        if messagebox.askyesno("Logout", "Are you sure you want to log out?"):
            dash.destroy()
            login_open_func()
    tk.Button(dash, text="Book", width=30, height=2, bg="#00bcd4", fg="white", command=book).pack(pady=10)
    tk.Button(dash, text="View Requests", width=30, height=2, bg="#00bcd4", fg="white", command=view_requests).pack(pady=10)
    tk.Button(dash, text="Logout", width=20, bg="#ff4d4d", fg="white", command=logout).pack(pady=20)


# ==============
# LOGIN SYSTEM
# ==============
def start_login_app():
    global root
    root = tk.Tk()
    root.title("Thunder Login")
    root.geometry("400x350")
    root.configure(bg="#1c1c28")

    #current mode
    current_mode = "student"
    
    def switch_mode(mode):
        nonlocal current_mode
        current_mode = mode
        if mode == "student":
            student_btn.config(bg="#00bcd4")
            admin_btn.config(bg="gray")
            student_frame.pack(pady=10)
            admin_frame.pack_forget()
        else:
            admin_btn.config(bg="#00bcd4")
            student_btn.config(bg="gray")
            admin_frame.pack(pady=10)
            student_frame.pack_forget()

    def login():
        if current_mode == "student":
            email = student_email.get().strip()
            password = student_password.get().strip()

            if not email.endswith('@pwr.nu.edu.pk'):
                messagebox.showerror("Error", "Invalid student email format.")
                return
            if password != "fast":
                messagebox.showerror("Error", "Incorrect password.")
                return

            student_id = email.split('@')[0]
            root.withdraw()
            show_student_dashboard(student_id, start_login_app)
            root.update()

        elif current_mode == "admin":
            email = admin_email.get().strip()
            password = admin_password.get().strip()

            if not email.endswith('@nu.edu.pk'):
                messagebox.showerror("Error", "Invalid admin email format.")
                return
            if password != "test":
                messagebox.showerror("Error", "Incorrect password.")
                return
            root.withdraw()
            show_admin_dashboard(start_login_app)
            root.update()

    #mode selector 
    mode_frame = tk.Frame(root, bg="#1c1c28")
    mode_frame.pack(pady=10)
    student_btn = tk.Button(mode_frame, text="Student Mode", width=15, command=lambda: switch_mode("student"), bg="#00bcd4", fg="white")
    student_btn.pack(side="left", padx=5)

    admin_btn = tk.Button(mode_frame, text="Admin Mode", width=15, command=lambda: switch_mode("admin"), bg="gray", fg="white")
    admin_btn.pack(side="left", padx=5)

    student_frame = tk.Frame(root, bg="#1c1c28")
    tk.Label(student_frame, text="Email (e.g., P233029@pwr.nu.edu.pk)", bg="#1c1c28", fg="white").pack()
    student_email = tk.Entry(student_frame, width=30)
    student_email.pack()
    tk.Label(student_frame, text="Password", bg="#1c1c28", fg="white").pack()
    student_password = tk.Entry(student_frame, width=30, show="*")
    student_password.pack()
    student_frame.pack(pady=10)

    admin_frame = tk.Frame(root, bg="#1c1c28")
    tk.Label(admin_frame, text="Email (e.g., admin@nu.edu.pk)", bg="#1c1c28", fg="white").pack()
    admin_email = tk.Entry(admin_frame, width=30)
    admin_email.pack()
    tk.Label(admin_frame, text="Password", bg="#1c1c28", fg="white").pack()
    admin_password = tk.Entry(admin_frame, width=30, show="*")
    admin_password.pack()

    login_btn = tk.Button(root, text="Login", width=20, bg="#00bcd4", fg="white", command=login)
    login_btn.pack(pady=15)

    root.mainloop()

display("Running login app... (Check pop-up window)")
start_login_app()
```
